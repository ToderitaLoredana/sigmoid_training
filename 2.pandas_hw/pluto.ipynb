```python
import pandas as pd
import numbers
```

```python
df = pd.read_csv('train.csv')
```

```python
df.info()
df.describe()
```

```python
# mean
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        print(f"{col}", df[col].mean())
```

```python
# median
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        print(f"{col}", df[col].median())

```

```python
# mode
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        print(f"{col}", df[col].mode())
```

```python
# standard deviation
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        print(f"{col}", df[col].std())
```

```python
# variance
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        print(f"{col}", df[col].var())
```

```python
# range of values
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        range_of_values = df[col].max() - df[col].min()
        print(f"{col}", range_of_values)
```
